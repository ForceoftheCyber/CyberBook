# Equilibrium Analysis

Before we can design a controller we need to analyze how the system behaves around an equilibrium point. An equilibrium point is a point where the time derivative of the system is 0, and it can be classified as being either **stable** or **unstable**. For the system above, the equilibrium point can be found by letting $x  = \theta = 0$, and then vary $z$.
